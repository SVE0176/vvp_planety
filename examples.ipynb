## Vliv délky časového kroku-Simulace je spuštěna pro různé délky časového kroku. S větším časovým krokem dochází k degradaci simulace — trajektorie vnějších planet (Jupiter, Saturn, Uran, Neptun) se začínají odchylovat od očekávaných eliptických drah.


```python
import matplotlib.pyplot as plt
from planety import nacti_planety, simulace, vykresli_trajektorie

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
nazvy = ["1 hodina (dt=3600)", "1 den (dt=86400)", "1 tyden (dt=604800)"]
kroky = [365*24*10, 365*10, 52*10]
dts = [3600, 86400, 604800]

for i, (dt, krok, nazev) in enumerate(zip(dts, kroky, nazvy)):
    planety = nacti_planety("data/planets.json")  # vzdy nacist znovu - ciste pocatecni podminky
    hist = simulace(planety, dt=dt, kroky=krok)
    for jmeno, polohy in hist.items():
        x = [p[0] for p in polohy]
        y = [p[1] for p in polohy]
        axes[i].plot(x, y, label=jmeno)
    axes[i].set_title(nazev)
    axes[i].set_xlabel("x [m]")
    axes[i].set_ylabel("y [m]")

plt.tight_layout()
plt.show()
```
